```python
import networkx as nx
import itertools
import pandas as pd
```

```python
chosen_disease = "lymphoma"

associated_proteins = {}

for modality in ['lipid', 'ion', 'protein', 'small_molecule', 'nucleic_acid']:
    graph_path = f"/n/holylabs/LABS/mzitnik_lab/Users/afang/ATOMICA/case_studies/atomica_net/graphs/ATOMICANet_{modality}.gml"
    graph = nx.read_gml(graph_path)

    nodes_in_disease = []
    for node in graph.nodes():
        diseases = graph.nodes[node]['diseases'].split(", ")
        if chosen_disease in diseases:
            nodes_in_disease.append(node)
    associated_proteins[modality] = [graph.nodes[node]['protein_name'] for node in nodes_in_disease]
```

In total there are 296 proteins associated with lymphoma from opentargets using our cutoffs. 281 are present in the network.

```python
num_unique_proteins = len(set(sum(associated_proteins.values(), [])))
num_unique_proteins
```

```python
modalities = ['lipid', 'ion', 'protein', 'small_molecule', 'nucleic_acid']
all_common_proteins = []
for modality1 in modalities:
    for modality2 in modalities:
        common_proteins = set(associated_proteins[modality1]).intersection(set(associated_proteins[modality2]))
        jaccard_similarity = len(common_proteins) / len(set(associated_proteins[modality1]).union(set(associated_proteins[modality2])))
        all_common_proteins.append([modality1, modality2, len(common_proteins), jaccard_similarity])

all_common_proteins = pd.DataFrame(all_common_proteins, columns=['modality1', 'modality2', 'num_common_proteins', 'jaccard_similarity'])
average_jaccard_similarity = all_common_proteins[all_common_proteins['modality1'] != all_common_proteins['modality2']]['jaccard_similarity'].mean()
print(f"Average Jaccard similarity between modalities: {average_jaccard_similarity}")
all_common_proteins.pivot(index='modality1', columns='modality2', values='num_common_proteins')
```
