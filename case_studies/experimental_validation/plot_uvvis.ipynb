## Candidate 1 - A0A7W1B5T5, HEC
binds

```python
control_data = pd.read_csv('/n/holylabs/LABS/mzitnik_lab/Users/afang/ATOMICA/case_studies/experimental_validation/UV-Vis/BSAcontrol_5min.csv')
control_data.dropna(inplace=True)
control_data['Heme only - Blank'] = control_data['Heme only'] - control_data['Blank']
control_data['Heme + protein - Blank'] = control_data['Heme + protein'] - control_data['Blank']

test_data = pd.read_csv('/n/holylabs/LABS/mzitnik_lab/Users/afang/ATOMICA/case_studies/experimental_validation/UV-Vis/Can1Recomb_5min.csv')
test_data.dropna(inplace=True)
test_data['Heme only - Blank'] = test_data['Heme only'] - test_data['Blank']
test_data['Heme + protein - Blank'] = test_data['Heme + protein'] - test_data['Blank']

control_data = control_data[(control_data['Wavelength (nm)'] < 510) & (control_data['Wavelength (nm)'] > 290)]
control_max_absorbance = control_data['Heme + protein - Blank'].max()
control_max_lambda = control_data.loc[(control_data['Heme + protein - Blank'] == control_max_absorbance), 'Wavelength (nm)'].values[0]

test_data = test_data[(test_data['Wavelength (nm)'] < 510) & (test_data['Wavelength (nm)'] > 290)]
test_max_absorbance = test_data['Heme + protein - Blank'].max()
test_max_lambda = test_data.loc[(test_data['Heme + protein - Blank'] == test_max_absorbance), 'Wavelength (nm)'].values[0]


fig, ax1 = plt.subplots(figsize=(1.5,2))
sns.lineplot(data=test_data, x='Wavelength (nm)', y='Heme only - Blank', color='#808080', ax=ax1)
sns.lineplot(data=test_data, x='Wavelength (nm)', y='Heme + protein - Blank', color='#00C4C7', ax=ax1)
ax1.set_xlim(290, 510)
ax1.set_ylim(0.0, 1.5)
ax1.set_xticks([300, 400, 500])
ax1.set_yticks([0.0, 0.5, 1.0, 1.5])

ax2 = ax1.twinx()
ax2.set_ylim(0.0, 1.5)
ax2.set_yticks([0.0, 0.5, 1.0, 1.5])
sns.lineplot(data=control_data, x='Wavelength (nm)', y='Heme + protein - Blank', color="#DE6E6E", ax=ax2)

ax1.vlines(x=test_max_lambda, ymin=0, ymax=test_max_absorbance, color="black", linestyle="--", linewidth=0.5)
ax2.vlines(x=control_max_lambda, ymin=0, ymax=control_max_absorbance, color="black", linestyle="--", linewidth=0.5)
ax1.text(
    0.3, 0.75,                       # (x, y) in axes fraction units
    f"{control_max_lambda:.1f}",
    transform=ax1.transAxes,           # interpret coords in axes space
    ha='left', va='top',              # anchor the text’s left-top corner
    fontsize=5
)
ax1.text(
    0.47, 0.97,                       # (x, y) in axes fraction units
    f"{test_max_lambda:.1f}",
    transform=ax1.transAxes,           # interpret coords in axes space
    ha='left', va='top',              # anchor the text’s left-top corner
    fontsize=5
)

ax1.set_xlabel('Wavelength (nm)')
ax1.set_ylabel('Absorbance (AU)')
ax2.set_ylabel('BSA Absorbance (AU)')
ax2.spines['top'].set_visible(False)
ax1.spines['top'].set_visible(False)

plt.savefig("figures/A0A7W1B5T5.svg", format='svg', bbox_inches='tight')
plt.show()
```
