```python
import pandas as pd
import pickle
```

```python
for ligand in [
    'ADP', 'ATP', 'BCB', 'BCL', 'CHL', 'CIT', 'CLA', 'FAD', 'GDP', 'GTP', 'HEC', 'HEM', 'MYR', 'NAD', 'NAP', 'NDP', 'OLA', 'PLM'
]:
    split_df = []
    for split in ["train", "valid", "test"]:
        with open(f"/n/holylfs06/LABS/mzitnik_lab/Lab/afang/ATOMICA/atomica_ligand/data/PL/{ligand}/{ligand}_sequence_30_{split}.pkl", "rb") as f:
            data = pickle.load(f)
        ids = [x['id'] for x in data]
        split_df.append(pd.DataFrame({'split': split, 'id': ids}))
    split_df = pd.concat(split_df)
    split_df.to_csv(f"/n/holylfs06/LABS/mzitnik_lab/Lab/afang/ATOMICA/atomica_ligand/data/PL/{ligand}/{ligand}_ids_sequence_30_split.csv", index=False)
```
