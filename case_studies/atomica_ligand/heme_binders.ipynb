```python
import pandas as pd
from glob import glob
import re
import pickle
```

```python
heme_dataset_dir = "/n/holylfs06/LABS/mzitnik_lab/Lab/afang/ATOMICA/atomica_ligand/data/PL/HEM"
positives = 0
negatives = 0
for split in ["train", "valid", "test"]:
    with open(f"{heme_dataset_dir}/HEM_sequence_30_{split}.pkl", "rb") as f:
        data = pickle.load(f)
    for item in data:
        if item["label"]:
            positives += 1
        else:
            negatives += 1
print(f"HEM dataset: {positives} positives, {negatives} negatives, {positives / (positives + negatives):.2%} positives")
```

```python
hem_sequences = {}
with open('HEM_annotated.fasta', 'r') as f:
    lines = f.readlines()
for l in lines:
    if l.startswith('>'):
        seq_id = l.split("|")[1]
        hem_sequences[seq_id] = ''
    else:
        seq = l.strip()
        hem_sequences[seq_id] += seq

matches = []
for seq_id, seq in hem_sequences.items():
    match = re.search(r'C[a-zA-Z]{2}CH', seq)
    if match:
        matches.append((seq_id, match.start(), match.end()))
print("HEM sequences with CXXCH motif:", len(matches))
matches
```
