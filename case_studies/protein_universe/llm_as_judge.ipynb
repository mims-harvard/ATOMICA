```python
import re
```

```python
with open("/n/netscratch/mzitnik_lab/Lab/afang/job_status/run_gpt4_58632359.out", "r") as f:
    results = f.readlines()
```

```python
def extract_answer_pairs(lines):
    # Regular expression pattern to capture numbers after specific phrases
    pattern = r"Correct answer:\s*(\d+),\s*Final answer:\s*(\d+)"
    
    # List to store the extracted integer pairs as tuples
    extracted_tuples = []
    
    for line in lines:
        # Use re.search() to find a match in each line
        match = re.search(pattern, line)
        if match:
            # Extract the integers using match groups
            correct_answer = int(match.group(1))
            final_answer = int(match.group(2))
            # Append the tuple of integers to the list
            extracted_tuples.append((correct_answer, final_answer))
    
    return extracted_tuples
```

```python
answer_pairs = extract_answer_pairs(results)
correct_answer = [x==y for x,y in answer_pairs]
print(f"Percentage of correct answers: {sum(correct_answer)/len(correct_answer)*100:.2f}%, {sum(correct_answer)} out of {len(correct_answer)}")
```
