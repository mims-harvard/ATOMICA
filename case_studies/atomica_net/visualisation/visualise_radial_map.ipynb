```python
import numpy as np
import pandas as pd
import holoviews as hv
import seaborn as sns
from holoviews import opts
hv.extension('bokeh')
```

```python
opts.defaults(opts.HeatMap(radial=True, width=800, height=800, tools=["hover"]))
```

```python
df = pd.read_csv("lymphoma_random_walk.csv")[['gene', 'count', 'modality']]
df = df[df['count'] != -1]
df['gene'] = df['gene'].apply(lambda x: x.split(' ')[0])
```

```python
len(df['gene'].unique())
```

```python
hv.extension('bokeh')
heatmap = hv.HeatMap(df, ["gene", "modality"])
heatmap.opts(opts.HeatMap(cmap=sns.light_palette("#4EBFE4", as_cmap=True), radial=True, yticks=5,
                        fontsize={'xticks': 20, 'yticks': 20}))
```

```python
hv.extension('matplotlib')
colors = {
    "ion": "#3CB37F",
    "nucleic_acid": "#EE8F78",
    "lipid": "#F5B856",
    "protein": "#D86ECC",
    "small_molecule": "#4EBFE4",
}
hv.extension('matplotlib')

for modality, color in colors.items():
    heatmap = hv.HeatMap(df, ["gene", "modality"])
    heatmap.opts(opts.HeatMap(cmap=sns.light_palette(colors[modality], as_cmap=True), radial=True, yticks=5,
                          xticks=len(df['gene'].unique()), fontsize={'xticks': 20, 'yticks': 20}))
    hv.save(heatmap, f"lymphoma_random_walk/lymphoma_random_walk_{modality}.svg", fmt="svg", size=800)
```

```python
heatmap = hv.HeatMap(df, ["gene", "modality"])
heatmap.opts(opts.HeatMap(cmap=sns.light_palette(colors['lipid'], as_cmap=True), radial=True, yticks=5,
                        fontsize={'xticks': 20, 'yticks': 20}, fig_size=400))
```
