```python
import shutil
import os
```

```python
def copy_files(ckpt_path, new_dir):
    weights_path = ckpt_path.replace(".ckpt", ".pt")
    config_path = os.path.dirname(ckpt_path) + "/config.json"
    version_path = os.path.dirname(os.path.dirname(ckpt_path))
    train_config = os.path.join(version_path, "train_config.json")
    args_path = os.path.join(version_path, "args.json")

    version_name = os.path.basename(version_path)
    new_dir = os.path.join(new_dir, version_name)
    os.makedirs(new_dir, exist_ok=True)

    print("Copying files to", new_dir)

    shutil.copy(ckpt_path, os.path.join(new_dir, os.path.basename(ckpt_path)))
    shutil.copy(weights_path, os.path.join(new_dir, os.path.basename(weights_path)))
    shutil.copy(config_path, os.path.join(new_dir, "config.json"))
    shutil.copy(train_config, os.path.join(new_dir, "train_config.json"))
    shutil.copy(args_path, os.path.join(new_dir, "args.json"))
```
