```python
import pandas as pd
import os
import pickle
```

```python
prediction_df = []
prediction_dir = "/n/holylfs06/LABS/mzitnik_lab/Lab/afang/protein_universe/foldseek_cluster/processed_A0A6B2YTG2/"
for fname in os.listdir(prediction_dir):
    if fname.endswith(".csv") and 'prediction' in fname:
        df = pd.read_csv(os.path.join(prediction_dir, fname))
        df['ligand'] = fname.split('_')[0]
        prediction_df.append(df)
prediction_df = pd.concat(prediction_df)
```

```python
prediction_df.groupby(['id', 'ligand']).agg({'predicted': 'mean', 'predicted': 'std'})
```
